# STA2002 Coursework 1: Regional Sales Analysis

This notebook analyses monthly sales for the North and South regions in 2025.

```python
import pandas as pd
import matplotlib.pyplot as plt
from scipy import stats
```

## Load the data

```python
df = pd.read_excel("sales_data.xlsx")
df
```

## Descriptive statistics

```python
summary = df[["North", "South"]].agg(["mean", "median", "std"])
summary
```

## Comparing the regions

```python
t_stat, p_value = stats.ttest_ind(df["North"], df["South"])
print(f"t = {t_stat:.3f}, p = {p_value:.2e}")
print(f"Difference in means = {df['North'].mean() - df['South'].mean():.2f}")
```

## Chart

```python
means = df[["North", "South"]].mean()
plt.bar(means.index, means.values, color=["steelblue", "darkorange"])
plt.show()
```

## Interpretation

### Descriptive statistics
Over the twelve months of 2025 the North region sold an average of 102.37 thousand units per month (median 101.5, standard deviation 4.91), while the South region averaged 92.50 thousand units (median 92.5, standard deviation 3.90). The North therefore outperformed the South in a typical month, and its sales were slightly more variable from month to month. In both regions the mean and median are close together, which suggests that monthly sales are not strongly skewed by a small number of unusually good or bad months.

### Comparing the regions
An independent-samples t-test was used because the two regions are separate groups of monthly observations. The test gave t = 5.069 with 22 degrees of freedom and p = 4.46e-05. Because the p-value is far below the significance level of 0.05, the difference in mean monthly sales between the regions is statistically significant. The North sold about 9.17 thousand more units per month than the South on average.

### What this means for the retailer
The result indicates that the gap between the regions is unlikely to be due to chance alone. A consistent difference of roughly nine thousand units per month is also practically important: over a year it amounts to more than 100 thousand units. Management could investigate whether factors such as store density, local marketing or product range explain the stronger performance of the North, and whether any of these could be transferred to the South.

### Limitations
The analysis uses only twelve monthly observations per region, so the estimates of the means and standard deviations are based on a small sample. The data cover a single year, so seasonal or one-off effects in 2025 may not be repeated in other years. The analysis also does not account for other variables, such as the number of stores or promotional activity in each region, so the difference cannot be interpreted as causal.
